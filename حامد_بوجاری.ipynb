Importing libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Reading the dataset
df = pd.read_csv('/content/coffe.csv')  # اسم فایل خودت رو اینجا بذار
```

```python
# Displaying first few rows
print(df.head())
```

```python
# Basic info about the dataset
print(df.info())
print(df.describe())
```

```python
# Checking for missing values
print(df.isnull().sum())
```

```python
# Example: simple analysis on a column (change 'ColumnName' to your column)
print(df['money'].value_counts())
```

```python
# Example: basic plot
df['money'].value_counts().plot(kind='bar')
plt.title('Distribution of Genre')
plt.xlabel('Categories')
plt.ylabel('Counts')
plt.show()
```
